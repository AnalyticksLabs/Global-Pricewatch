## Handling Missing Values

```python
df.isnull().sum().sum()
```

```python
df.isnull().sum()
```

```python
df.dropna(axis=1, how="all", inplace=True)
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
missing_meta = df[
    df.isnull().any(axis=1)
]

print(missing_meta)
```

```python
df.drop(
    columns=["Missing_Count"],
    errors="ignore",
    inplace=True
)

# Save cleaned metadata
df.to_csv(
    "META_1.csv",
    index=False
)

print("Saved successfully!")
print(df.shape)
print("Duplicates:", df.duplicated().sum())
```
