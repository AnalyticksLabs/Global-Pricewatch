```python
import pandas as pd
```

```python
df = pd.read_csv("Global Dataset of Inflation.csv", encoding="latin1")
```

```python
df.head()
```

# Data Cleaning 


## Handling Missing Values


```python
df.isnull().sum().sum()
```

```python
df = df.dropna(axis=1, how="all")
```

```python
df.head(5)
```

```python
df.isnull().sum().sum()
```

```python
df.tail(20)
```

```python
year_cols = df.columns[5:]

df["missing_pct"] = df[year_cols].isna().mean(axis=1) * 100

df[["Country", "missing_pct"]].sort_values("missing_pct", ascending=False)
```

```python
df = df[df["missing_pct"] <= 30].copy()
```

```python
year_cols = df.columns[5:]

df[year_cols] = df[year_cols].apply(pd.to_numeric, errors='coerce')
```

```python
df[year_cols] = df[year_cols].interpolate(
    axis=1,
    limit=2,
    limit_area="inside"
)
```

```python
df.isnull().sum().sum()
```

## Handling Duplicate Values

```python
df.duplicated().sum()
```

## Finishing up

```python
year_cols = df.columns[5:]

print(df[year_cols].min().min())
print(df[year_cols].max().max())
```

```python
xyear_cols = df.columns[5:]

# Find maximum value
max_val = df[year_cols].max().max()
print("Maximum inflation:", max_val)

# Find where it occurs
for col in year_cols:
    rows = df[df[col] == max_val]
    if not rows.empty:
        print(rows[['Country', col]])
```

```python
min_val = df[year_cols].min().min()
print("Minimum inflation:", min_val)

for col in year_cols:
    rows = df[df[col] == min_val]
    if not rows.empty:
        print(rows[['Country', col]])

        
```

```python
df.to_csv("Global_Dataset_of_Inflation_Cleaned.csv", index=False)
```
