# Data Cleaning 


## Handling Missing values

```python
df.isnull().sum()
```

```python
df.iloc[:, 2:] = df.iloc[:, 2:].interpolate(axis=1, limit_direction='both')
```

```python
df.head()
```

```python
df.isnull().sum()
```

## Handling Duplicate Values


```python
df.drop_duplicates(inplace=True)
```
