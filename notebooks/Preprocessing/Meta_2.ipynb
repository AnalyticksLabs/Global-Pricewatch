# Cleaning

## Minor Cleaning of column names

```python
indicator_meta.columns = (
    indicator_meta.columns
    .str.strip()
    .str.replace('"', '', regex=False)
)
```

## Dropping Missing Values

```python
indicator_meta.dropna(axis=1, how="all", inplace=True)
```

## Dropping Duplicates

```python
indicator_meta.drop_duplicates(inplace=True)
```
